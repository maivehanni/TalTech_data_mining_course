## 2.1. Stringid, jadad ja tsüklid

Andmete eeltöötlus on oluline osa andmekaeve protsessist, mis hõlmab andmete lugemist erinevatest failidest või andmebaasidest, nende teisendamist sobivale kujule, vigaste andmete eemaldamist jne. ### Stringid

String (ehk sõne) on tekstijada ehk märgijada. String on Pythonis jutumärkide vahel ('abc', "abc"). Stringidele võib rakendada ka operatsioone + (liitmine omavahel), * (korrutamine arvuga).

```python
s1 = "Andme"
s2 = 'kaeve'
aine = s1 + s2
aine
```

```python
"1"+"2"
```

```python
s1*3
```

Stringide vormindamine tähendab väärtuse sisestamist mallstringi. Siin me vaatleme % notatsiooni. On olemas ka uuem [format()](https://docs.python.org/3/library/string.html#formatstrings) meetodi notatsioon. Sisestatavat väärtust (lünka) mallstringis märgivad sümbolid %s (string), %f (float reaalarvud), %d (double reaalarv), %i (integer täisarv). Notatsioon `mallstring % (v1, v2,...)` täidab lüngad sobivat tüüpi väärtustega v1, v2.

```python
mall = "Tere %s!"
print(mall % 'talv')
objekti_mall = "Nimi: %s, Vanus: %i"
print(objekti_mall % ("Mr. X", 33))
print(objekti_mall % ("Ms. Z", 23))
```

```python
objekti_mall
```

Sama asi `format()` kujul:

```python
objekti_mall = "Nimi: {0}, Vanus: {1}"
print(objekti_mall.format("Mr. X", 33))
print(objekti_mall.format("Ms. Z", 23))
```

Võib kasutada ka nimedega vormindusargumente:

```python
objekti_mall = "Nimi: {nimi}, Vanus: {vanus}"
print(objekti_mall.format(nimi="Mr. X", vanus=33))
print(objekti_mall.format(nimi="Ms. Z", vanus=23))
```

Selline kuju on tihti mugav väljundi vormindamiseks kui meil on vaja väljastada mitu väärtust.

### Dictionary (sõnastik)

[Dictionary (sõnastik)](https://docs.python.org/3/tutorial/datastructures.html#dictionaries) indekseerib elemente, mitte täisarvuliste indeksitega *0, 1,...,n* vaid kasutaja määratud võtmetega. Sõnastiku elemendid asuvad loogeliste sulgude vahel kujul `{võti_1: element_1, võti_2: element_2,...}`. Matemaatikas vastab sõnastikule [kujutise](https://et.wikipedia.org/wiki/Kujutis_(matemaatika)) mõiste.

```python
d = {"petal_l": [1, 2, 3], "sepal_l": [2, 2, 10], "sepal_w": [3, 5, 2]}
print("Sõnastik alguses:", d)
d["petal_l"] = [0, 0, 7]
print("Tsükkel algab")
for key in d:
    print(key)
    print(d[key])
print("Tsükkel lõpeb")
print(d.get("puudu", 0))
```

```python
try:
    d["puudu"]
except KeyError:
    print("Võtme puudumine annab d[võti] päringu korral veateate!")
```

Mõned olulisemad sõnastiku meetodid ja muud operatsioonid:
* `d[v] = x` lisab võtme `v` alla väärtuse `x`

* `d.items()` tagastab `(võti, väärtus)` ennikute jada
* `del d[v]` kustutab sõnastiku võtmega *v* määratud elemendi 
* `for v in d` itereerib üle kõigi sõnastiku `d` võtmete `v`
* `d.get(v, default)` tagastab võtme `v` olemasolul sellega määratud väärtuse, muidu `default`.
* `d.setdefault(v, default)` tagastab võtme `v` olemasolul sellega määratud väärtuse, muidu väärtustab võtme v väärtusega `default` ja tagastab selle väärtuse.

### Failitöötlus

[Failidest](https://docs.python.org/3/tutorial/inputoutput.html#reading-and-writing-files) andmete lugemine ja nendesse andmete kirjutamine on oluline osa andmete eeltöötlusest. Tavaliselt töötame me *.csv* formaadis failidega, mille jaoks on erifunktsioonid [pandas](https://pandas.pydata.org/pandas-docs/version/0.23.4/10min.html) moodulis, aga muude juhtude tarvis on vaja tunda üldist funktsionaalsust.


Faili avab ja failiobjekti tagastab *open(filename, mode)* standardfunktsioon. 
 *Mode* režiim võib olla 'r' ainult lugemiseks, 'w' ainult kirjutamiseks (olemasolev fail kirjutatakse üle),  'a' lisamiseks olemasoleva faili lõppu, 'r+' nii lugemiseks kui kirjutamiseks. 
Faili sulgemiseks tuleb kasutada tagastatud failiobjekti *close()* meetodit.

```python
f = open("helloworld.txt", "w")
f.write("Hello world!")
f.close()
```

Üldiselt soovitatakse tööks failidega kasutada **with** `object` **as** *name*: plokki, mis suleb failiobjekti automaatselt, ka veaolukorra ilmnemisel.

```python
with open("helloworld.txt", "a") as f:
    for i in range(10):
        f.write("Hello "*i  + "World!\n")
```

```python
with open("helloworld.txt", "r") as f:
    filetext = f.readlines()
print(filetext)
```

```python
print(filetext[:5])
```

```python
# Näide sõnastiku setdefault meetodi kasutamisest sageduste arvutamisel
with open("helloworld.txt", "r") as hellofile:
    fulltext = hellofile.read()
freq_dict = {}
for word in fulltext.split():
    freq_dict[word] = freq_dict.setdefault(word, 0) + 1
freq_dict
```

Olulisemad meetodid tööks tekstifailidega on 
* `f.read()`, mis loeb faili tervikteksti ühe stringina, 
* `f.readlines()`, mis tagastab stringide listi, mille elemendid on faili read,
* `f.readline()` loeb failist f ühe rea ja liigutab järjekorda failis ühe rea võrra edasi,
* `f.write(s)`, mis kirjutab faili stringi *s* ( reavahetuseks kasutada `'\n'` erisümbolit).

### List comprehension (loendi hõlmamine)

[List comprehension (loendi hõlmamine)](https://docs.python.org/3/tutorial/datastructures.html#list-comprehensions) võimaldab genereerida loendeid, sõnastikke ja ennikuid ilma abimuutujaid kasutamata. Lihtne kuju on `[avaldis_elemendi_baasil for element in list]`, mis teisendab listi elemendid vastavalt avaldisele. Eesmärgiks on asendada tavaline tsükkel lihtsama avaldisega, kus uue loendi tekitamist ja sinna sammhaaval elementide lisamist ei ole vaja programmis ilmutatult kirja panna. Selline funktsionaalne ja deklaratiivne lähenemine on kasutusel ka andmeteaduse andmestruktuuride juures ($numpy.array, pandas.Dataframe$), kus for-tsükleid praktiliselt ei kasutata.

```python
# Tavaline tsükkel
l =  []
for i in range(10):
    l.append(2**i)
l
```

```python
# Ekvivalentne list comprehension
l = [2**i for i in range(10)]
l
```

*List comprehension* võimaldab lihtsat loendi elementide teisendust ja filtreerimist.

```python
arvustr = "0.9, 0.5, 0.4, 0.2"
# Tahame selle teisendada reaalarvude listiks
# print("Stringide list:", arvustr.split(","))
# Tahame teisendada iga selle listi elemendi reaalarvuks
arvulist = [float(e) for e in arvustr.split(",")]
print("Arvude list:", arvulist)
```

```python
# Ekvivalentne kuju for tsükli baasil
arvulist = []
for e in arvustr.split(","):
    arvulist.append(float(e))
    
print("Arvude list:", arvulist)    
    
```

Üldine kuju on:
\[*teisendusavaldis_muutujaga* **for** *muutuja* **in** *jada* **if** *filtreerimisavaldis_muutujaga*\], mis teisendab listi elemendid vastavalt avaldisele ja filtreerib vastavalt tingimusele.

```python
from random import random
random()
```

```python
rea_pikkused = [len(line) for line in filetext if random()>0.5]
rea_pikkused
```

```python
arvulist = [float(e) for e in arvustr.split(",") if float(e)>=0.5]
arvulist
```

Loendi hõlmamine sarnaneb hulga defineerimisega hulgateoorias. Näiteks hulga S põhjal saame defineerida uue hulga kujul:

$$ \{2t+1 \mid t \in S, min<t<max) \} $$

*(Võta kõik elemendid t hulgast S, mille puhul min<t<max on tõene ja teisenda need elementideks 2t+1)*

Vastav loendi hõlmamine oleks:

`[2t+1 for t in S if (min<t<max)]`


Tabeli esitus listidena oleks listide list, kus kõik sisemised listid on sama pikkusega.

```python
korrutustabel = [[x*y for x in range(1, 10)] for y in range(1, 10)]
korrutustabel
```

```python
# Ligipääs reale on tabel[rida]
korrutustabel[8]
```

```python
# Ligipääs 'lahtrile' on tabel[rida][veerg]
korrutustabel[6-1][7-1]
```

Loendi hõlmamist saab loomulikult rakendada ka tekstile ja failidele. Leiame kahekümnest märgist lühemate ridade arvu faili tekstist $$|\{r \mid r \in f, |r|<20\}|$$ (püstkriipsud hulga ümber näitavad hulga elementide arvu, Pythonis on selleks funktsioon $len()$).

```python
len([r for r in filetext if len(r) < 20])
```

Lisaks loendi hõlmamisele on võimalik loogelisi sulge kasutades defineerida ka sõnastike hõlmamisi. Siin tuleb määrata nii võti kui väärtus kujul *võti:väärtus*.

```python
{i:2**i for i in range(10)}
```
